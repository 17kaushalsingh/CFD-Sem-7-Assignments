## Plots for CFD Assignment

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('final_profile.csv')
df.head()
```

```python
x = df.columns.astype(float)
T = df.iloc[0].values
```

```python
plt.figure(figsize=(8, 5))
plt.plot(x, T, marker='o')
plt.xlabel('x (m)')
plt.ylabel('T (°C)')
plt.title('Temperature Profile (T vs x)')
plt.grid(True)
plt.show()
```
